```python
import numpy as np
import torch
import torch.nn as nn # used for training and building neural networks
import torch.optim as optim # for otpimizers
```

# Introduction

## Training Loops

```python
# Data
xs = torch.tensor(np.arange(50).reshape(50, 1), dtype=torch.float32)
ys = 2 * xs + 1

# model
# sequential model, input goes thru layers sequentially
# model has only one layer, where it accepts an input of size 1
# and outputs a size 1 tensor
model = nn.Sequential(nn.Linear(1, 1))
# Linear layer so its a y = np.dot(weights.T, xs) + b

# Loss function
criterion = nn.MSELoss() # mean square error

# optimizer
# choose an optimizer then within it always put model.parameters()
optimizer = optim.SGD(model.parameters(), lr = 1e-5)
# the optimizer updates the weights 
# in the direction of the gradient vector

# for using the gpu
device = torch.device("cuda:0")
model = model.to(device)

# Training Loop
for _ in range(500):
    model.train()

    # moves dataset to gpu
    xs = xs.to(device)
    ys = ys.to(device)
    
    optimizer.zero_grad() # this zeros the gradient for every loop
    
    outputs = model(xs)
    loss = criterion(outputs, ys)
    print(f"Epoch {_}: {loss}")
    
    loss.backward() 
    # this finds the gradient vector, 
    # partial derivative of loss_fn with respect to weights
    
    optimizer.step()
```

## Evaluation

```python
model.eval()
x_k = torch.tensor([[-1], [100]], dtype = torch.float32).to(device)
with torch.no_grad():
    y_hat = model(x_k)
    print(y_hat)
```
