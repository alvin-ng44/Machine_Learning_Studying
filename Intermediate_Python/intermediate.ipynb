# String Concatentation and Formatting

```python
names = ["Alvin", "Simon", "Theodore"]

for name in names:
    statement = " ".join(["Hello", name + "!"])
    print(statement)

# " ".join(lst)
# gives lst[0] + " " + lst[1] + " " + ... + lst[n]
# basically a space inbetween elements of the list
```

```python
lst0 = ["Alvin", "Simon", "Theodore"]
lst1 = [20, 10 ,5]
lst2 = ["apples", "oranges", "pears"]

for lst in list(zip(lst0, lst1, lst2)):
    statement = " ".join([lst[0], "has", str(lst[1]), f"{lst[2]}."])
    print(statement)

```

```python
who = "Gary"
how_many =  10
what = "apples"

print(f"{who} has {how_many} {what}.")
print("{} has {} {}.".format(who, how_many, what))
```
