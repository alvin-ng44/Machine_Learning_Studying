# String Concatentation and Formatting

```python
names = ["Alvin", "Simon", "Theodore"]

for name in names:
    statement = " ".join(["Hello", name + "!"])
    print(statement)

# " ".join(lst)
# gives lst[0] + " " + lst[1] + " " + ... + lst[n]
# basically a space inbetween elements of the list
```

```python
lst0 = ["Alvin", "Simon", "Theodore"]
lst1 = [20, 10 ,5]
lst2 = ["apples", "oranges", "pears"]

for lst in list(zip(lst0, lst1, lst2)):
    statement = " ".join([lst[0], "has", str(lst[1]), f"{lst[2]}."])
    print(statement)

```

```python
who = "Gary"
how_many =  10
what = "apples"

print(f"{who} has {how_many} {what}.")
print("{} has {} {}.".format(who, how_many, what))
```

# List comprehension and generator expressions

```python
# Basic example of list comprehension
[i for i in range(10)]
[i**2 for i in range(10)]
```

```python
def is_even(num):
    return num % 2 == 0

# List comprehension with if statement
[i for i in range(10) if is_even(i)]
```

```python
for i in range(1, 4):
    for ii in range(1, 6):
        print(i, i* ii)
```

```python
# this is equivalent to
[[print(i, i * ii) for ii in range (1, 6)] for i in range(1, 4)]

# the inner loop 
```
