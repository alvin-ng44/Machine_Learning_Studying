```python
import numpy as np
```

```python
# Suppose we have 4 inputs and 1 neuron. 
# There would be 4 weights (from 4 inputs to 1 neuron)
# and 1 bias (for the neuron)

inputs = [1, 2, 3, 2.5]
weights = [0.2, 0.8, -0.5, 0.1]
bias = 2

# formula for the neuron is: np.dot(inputs, weights) + bias

neuron_output = 0
for input, weight in zip(inputs, weights):
    neuron_output += input * weight
neuron_output += bias
print(neuron_output)
```

```python
# Suppose we have 4 inputs and 3 neurons. 
# So there would be 4 weights for each neuron (one from each input to neuron)
# 3 biases (one for each neuron)

inputs = [1, 2, 3, 2.5]

# weights for 3 neurons (4 weights for each neuron)
weights = [
    [0.2, 0.8, -0.5, 1],
    [0.5, -0.91, 0.26, -0.5],
    [-0.26, -0.27, 0.17, 0.87]
]
biases = [2, 3, 0.5]

# for the layer of 3 neurons, we would have 3 outputs (one for each neuron)
# the inputs are the same for each neuron
# but the weights and biases are different
# so loop over weights and biases first (since for each neuron, we have different weights and bias)

# next, there is a edge connecting each input to the neuron
# so loop over inputs and neuron_weights

# then add the neuron_bias to the output of the neuron

layer_outputs = []
for neuron_weights, neuron_bias in zip(weights, biases):
    neuron_output = 0
    for input, weight in zip(inputs, neuron_weights):
        neuron_output += input * weight
    neuron_output += neuron_bias
    layer_outputs.append(neuron_output)

print(layer_outputs)
```

```python
# Equivalent code using numpy
# For a single neuron
inputs = [1, 2, 3, 2.5]
weights = [0.2, 0.8, -0.5, 0.1]
bias = 2

neuron_output = np.dot(inputs, weights) + bias
print(neuron_output)
```

```python
# Equivalent code using numpy
# for 4 inputs and a layer of 3 neurons

inputs = np.array([1, 2, 3, 2.5])
weights = np.array([
    [0.2, 0.8, -0.5, 1],
    [0.5, -0.91, 0.26, -0.5],
    [-0.26, -0.27, 0.17, 0.87]
])
biases = np.array([2, 3, 0.5])

layer_outputs = np.dot(weights, inputs) + biases
print(layer_outputs)

# if a is a 2D array and b is a 1D array
# it is a sum product over the last axis of a and b
# this means that it does the dot product of each row of a with b
```
