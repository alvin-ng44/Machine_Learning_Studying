```python
import torch
import numpy as np
```

# Creating tensors in PyTorch

```python
a = [1, 2, 3]
t_a = torch.tensor(a)
print(t_a)
```

```python
b = np.arange(4, 7, dtype = np.int32)
t_b = torch.from_numpy(b)
print(t_b)
```

```python
# there are equivalent functions of np in torch
# np.ones --> torch.ones
t_ones = torch.ones(size = (2,3))
print(t_ones)

# they also have properties such as dtype and shape
print(t_ones.dtype)
print(t_ones.shape)
```

```python
# Both are equivalent ways to get random samples from U(0,1)
# but rand_tensor0 can be used to get samples from U(r1, r2) too

rand_tensor0 = torch.FloatTensor(2, 3).uniform_(0, 1)
print(rand_tensor0)

rand_tensor1 = torch.rand(2, 3)
print(rand_tensor1)
```

```python
print(t_a)
print(t_a.dtype)
# change data type of a tensor
t_a_new = t_a.to(torch.float64)
print(t_a_new)
print(t_a_new.dtype)
```

## Useful operations

```python
# Transpose

t = torch.rand(size = (3, 5))
t_tr = t.T

print(t)
print(t_tr)

# another way to transpose for tensors of higher dimensions

# torch.transpose(tensor, dim0, dim1)
# swaps dim0 and dim1 of tensor
t_tr1 = torch.transpose(t, 0, 1)
print(t_tr1)
```

```python
t = torch.zeros(30)
t_reshape = t.reshape(5, 6)
print(f"{t.shape} --> {t_reshape.shape}")
```

```python
# to remove unwanted dims whose dims = 1
t = torch.zeros(size = (1, 5, 7, 1, 3, 4))
print(t.shape)

# this removes all dims of size 1
t_sqz = t.squeeze() 
print(t_sqz.shape)

# this tells which dim to squeeze
# in this case, this squeezes dim 0
# if u do torch.squeeze(t, 2)
# nothing happens since the second dim has size 7
t_sqz1 = torch.squeeze(t, 0) 
print(t_sqz1.shape)

t_sqz2 = torch.squeeze(t, 2)
print(t_sqz2.shape)
```
